```python
from deal_pilot.rag.store.feishu_doc.get_store import get_store

store = await get_store()
```

```python
question="你有没有A200的产品图？"
```

```python
result = await store.asimilarity_search_with_score(
    query=question,
    fetch_k=10,
    k=4,
    param=[
        {
            "metric_type":"COSINE",
            "params":{
                "ef":64,
                "radius":0.5
            }
        },
        {
            "metric_type":"BM25",
            "params":{}
        }
    ],
    expr='node_title like "%A200%"'
)
```

```python
result
```

```python
[
    {"doc":doc, "score":score}
    for doc, score in result
    if score >= 0.03
]
```

```python
from pymilvus import Function , FunctionType

rrf_reranker = Function(
    name="dense_sparse_rrf",
    input_field_names=[],
    function_type=FunctionType.RERANK,
    params={
        "reranker":"rrf",
        "k":60
    }
)
```

```python
await store.asimilarity_search_with_score(
    query=question,
    fetch_k=10,
    k=4,
    param=[
        {
            "metric_type":"COSINE",
            "params":{
                "ef":64,
                "radius":0.5
            }
        },
        {
            "metric_type":"BM25",
            "params":{}
        }
    ],
    expr='node_title like "%A200%"',
    reranker=rrf_reranker
)
```

```python
weighted_reranker = Function(
    name="dense_sparse_weighted",
    input_field_names=[],
    function_type=FunctionType.RERANK,
    params={
        "reranker":"weighted",
        "weights":[0.7,0.3],
        "norm_score":True
    }
)
```

```python
await store.asimilarity_search_with_score(
    query=question,
    fetch_k=10,
    k=4,
    param=[
        {
            "metric_type":"COSINE",
            "params":{
                "ef":64,
                "radius":0.5
            }
        },
        {
            "metric_type":"BM25",
            "params":{}
        }
    ],
    expr='node_title like "%A200%"',
    reranker=weighted_reranker
)
```
