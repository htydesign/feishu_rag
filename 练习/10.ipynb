```python
import paddle.inference as paddle_infer

# 规避 Paddle 3.x 在 Windows CPU 上的已知 Bug（PaddleOCR issue #18162）：
# 图优化会生成 fused_conv2d 融合算子，其内核错误地走 OneDNN 上下文导致报错，
# 关闭推理图优化即可绕过（enable_mkldnn=False 挡不住这个 Bug）。
_orig_create_predictor = paddle_infer.create_predictor
def _create_predictor_without_ir_fusion(config):
    config.switch_ir_optim(False)
    return _orig_create_predictor(config)
paddle_infer.create_predictor = _create_predictor_without_ir_fusion

from paddleocr import PaddleOCR

ocr = PaddleOCR(
    use_gpu=False,
    lang="ch",
    use_angle_cls=True,
    enable_mkldnn=False,
)
```

```python
results = ocr.ocr(path)
```
