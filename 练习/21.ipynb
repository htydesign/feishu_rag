```python
from pymilvus import AsyncMilvusClient

collection_name = "feishu_doc"
db_name="deal_pilot"
client = AsyncMilvusClient(
    uri="http://127.0.0.1:19530",
    db_name=db_name,
)
```

```python
await client.load_collection(collection_name=collection_name)
await client.get_load_state(collection_name=collection_name)
```

```python
rows = await client.query(
    collection_name=collection_name,
    filter='text like "%生成器%"',
    output_fields=["id","text"],
    limit = 5
)

from rich import print

for row in rows:
    print("id:",row["id"])
    print("text:",row["text"][:120])
    print("-"*60)
```

```python
from deal_pilot.rag.embedding_model import create_embedding_model

embedding_model = create_embedding_model()

question = "ES6中正则表达式有什么变动？"
query_vector = await embedding_model.aembed_query(question)

len(query_vector)
```

```python
results = await client.search(
    collection_name = collection_name,
    data=[query_vector],
    anns_field="vector",
    limit=5,
    search_params={
        "metric_type":"COSINE",
        "params":{"ef":60},
    },
    output_fields=["text"]
)

for rank, hit in enumerate(results[0], start=1):
    print(dict.keys(hit))
    print(f"第 {rank} 名，id={hit['id']}，COSINE={hit['distance']:.4f}")
    # print(hit["entity"]["text"])
    # print('-'*60)
```

```python
def show_result(results):
    for i, result in enumerate(results, start=1):
        print(f"问题{i}的检索结果：")
        for rank, hit in enumerate(result, start=1):
            print(dict.keys(hit))
            print(f"第 {rank} 名，id={hit['id']}，COSINE={hit['distance']:.4f}")
            print(hit["entity"]["text"])
            print('-'*60)
        print('='*60)
```

```python
show_result(results)
```

```python
filtered_results = await client.search(
    collection_name=collection_name,
    anns_field="vector",
    data=[query_vector],
    filter='text like "%正则%"',
    limit=5,
    output_fields=["text"],
)

show_result(filtered_results)
```

```python
range_results = await client.search(
    collection_name=collection_name,
    anns_field="vector",
    data=[query_vector],
    limit=5,
    search_params={
        "metric_type": "COSINE",
        "params": {"ef": 64, "radius": 0.55},
    },
    output_fields=["text"],
)

show_result(range_results)
```

```python
await client.close()
```
