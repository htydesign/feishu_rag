```python
from pymilvus import AsyncMilvusClient

collection_name = "feishu_doc"
client = AsyncMilvusClient(
    uri="http://127.0.0.1:19530",
    db_name="deal_pilot"
)
```

```python
import pypdf

pdf_path = "../resources/ECMA.pdf"

reader = pypdf.PdfReader(pdf_path)

chunks: list[str] = []

for page in reader.pages:
    chunks.append(page.extract_text())

len(chunks)
```

```python
from deal_pilot.rag.embedding_model import create_embedding_model

vectors = await create_embedding_model().aembed_documents(chunks)
len(vectors)
```

```python
entities = [
    {"text":text,"vector":vector} for text,vector in zip(chunks,vectors)
]
```

```python
len(entities)
```

```python
entities[100]
```

```python
await client.insert(
    collection_name=collection_name,
    data= entities
)
```

```python
await client.delete(
    collection_name=collection_name,
    ids=[470020153543789076]
)
```
